```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matplotlib

matplotlib.rcParams['font.sans-serif'] = ['SimHei']
matplotlib.rcParams['axes.unicode_minus'] = False

data = {
    "普通本科":[721,738,749,761,791],
    "中等职业教育": [620,601,593,582,557],
    "普通高中": [797,797,803,800,793]
}

df = pd.DataFrame(data, index=[2014, 2015, 2016, 2017, 2018])

plt.figure(figsize=(12, 8))

bar_width = 0.25
years = df.index
x_pos = np.arange(len(years))

bars1 = plt.bar(x_pos - bar_width, df['普通本科'], bar_width, 
                label='普通本科', color='skyblue', alpha=0.8)
bars2 = plt.bar(x_pos, df['中等职业教育'], bar_width, 
                label='中等职业教育', color='lightcoral', alpha=0.8)
bars3 = plt.bar(x_pos + bar_width, df['普通高中'], bar_width, 
                label='普通高中', color='lightgreen', alpha=0.8)

def add_value_labels(bars):
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 5,
                f'{int(height)}',
                ha='center', va='bottom', fontsize=10, fontweight='bold')

add_value_labels(bars1)
add_value_labels(bars2)
add_value_labels(bars3)

plt.title('2014-2018年三类学校报名人数统计', fontsize=16, fontweight='bold')
plt.xlabel('年份', fontsize=12)
plt.ylabel('报名人数（万人）', fontsize=12)
plt.xticks(x_pos, years)
plt.ylim(0, max(df.max()) + 50)
plt.legend(ncol=3, loc='upper left')
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()
print("数据表格：")
print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from datetime import datetime

df_bj = pd.read_csv('guazi_bj.csv')
df_gz = pd.read_csv('guazi_gz.csv') 
df_sh = pd.read_csv('guazi_sh.csv')
df_sz = pd.read_csv('guazi_sz.csv')

df_all = pd.concat([df_bj, df_gz, df_sh, df_sz], ignore_index=True)

def calculate_use_year(buy_time):
    if pd.isna(buy_time):
        return np.nan
    try:
        buy_date = pd.to_datetime(buy_time + "-01")
        current_date = datetime.now()
        use_years = (current_date - buy_date).days / 365.25
        return round(use_years, 1)
    except:
        return np.nan

df_all['use_year'] = df_all['buy_time'].apply(calculate_use_year)
df_all['hedge_rate'] = df_all['es_price'] / df_all['new_price']

df_clean = df_all.dropna(subset=['use_year', 'hedge_rate', 'km'])
df_clean = df_clean[(df_clean['hedge_rate'] > 0) & (df_clean['hedge_rate'] <= 1)]
df_clean = df_clean[(df_clean['use_year'] > 0) & (df_clean['use_year'] <= 15)]
df_clean = df_clean[(df_clean['km'] > 0) & (df_clean['km'] <= 25)]

plt.figure(figsize=(15, 6))

plt.subplot(1, 2, 1)
plt.scatter(df_clean['use_year'], df_clean['hedge_rate'], alpha=0.6, s=20)
plt.xlabel('use_year')
plt.ylabel('hedge_rate')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.scatter(df_clean['km'], df_clean['hedge_rate'], alpha=0.6, s=20)
plt.xlabel('km')
plt.ylabel('hedge_rate')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['font.sans-serif'] = ['SimHei']  
plt.rcParams['axes.unicode_minus'] = False    

labels = ['Chrome', 'Internet Explorer', 'Firefox', 'Edge', 'Safari', '其他浏览器']
sizes = [60.98, 12.18, 11.47, 4.15, 3.27, 7.95]  

explode = (0.05, 0, 0, 0, 0, 0) 

colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#ff99cc', '#c2c2f0']

fig, ax = plt.subplots(figsize=(10, 8))

wedges, texts, autotexts = ax.pie(sizes, 
                                  explode=explode,
                                  labels=labels, 
                                  colors=colors,
                                  autopct='%1.2f%%',
                                  shadow=True,  
                                  startangle=90)

for text in texts:
    text.set_color('black')
    
for i, autotext in enumerate(autotexts):
    autotext.set_color('white')
    autotext.set_weight('bold')
    
    if labels[i] in ['Edge', 'Safari']:
        autotext.set_fontsize(10)
    else:
        autotext.set_fontsize(12)

for i, text in enumerate(texts):
    if labels[i] in ['Edge', 'Safari']:
        text.set_fontsize(10)
    else:
        text.set_fontsize(12)

ax.axis('equal')
plt.title('浏览器市场份额分布', fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
print("浏览器市场份额数据:")
for label, size in zip(labels, sizes):
    print(f"{label}: {size}%")
print(f"总计: {sum(sizes)}%")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['font.sans-serif'] = ['SimHei']  
plt.rcParams['axes.unicode_minus'] = False    

df = pd.read_csv('scores.csv')

score_columns = [col for col in df.columns if col not in ['num', 'class']]

subject_means = []
for col in score_columns:
    mean_score = df[col].mean()
    subject_means.append(mean_score)

N = len(score_columns)
angles = [n / float(N) * 2 * np.pi for n in range(N)]
angles += angles[:1]  

values = subject_means + [subject_means[0]]

fig, ax = plt.subplots(figsize=(12, 12), subplot_kw=dict(projection='polar'))


colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FFEAA7', '#DDA0DD', '#98D8C8']
ax.plot(angles, values, 'o-', linewidth=3, color=colors[0], alpha=0.8, markersize=8)
ax.fill(angles, values, alpha=0.3, color=colors[0])

ax.set_xticks(angles[:-1])
ax.set_xticklabels(score_columns, fontsize=14, fontweight='bold')

ax.set_ylim(0, 100)
ax.set_yticks([20, 40, 60, 80, 100])
ax.set_yticklabels(['20', '40', '60', '80', '100'], fontsize=12, alpha=0.7)
ax.grid(True, alpha=0.3)

for angle, value, subject in zip(angles[:-1], subject_means, score_columns):
    ax.text(angle, value + 5, f'{value:.1f}', 
            horizontalalignment='center', verticalalignment='center',
            fontsize=12, fontweight='bold', 
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))

plt.title('各科目平均成绩雷达图', size=18, fontweight='bold', pad=40)

overall_mean = np.mean(subject_means)
mean_line = [overall_mean] * len(angles)
ax.plot(angles, mean_line, '--', linewidth=2, color='red', alpha=0.7, label=f'总平均分: {overall_mean:.1f}')
plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.0))
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

bj_temps = [29, 27, 23, 22]
bj_hours = ["20时", "23时", "2时", "5时"]

plt.figure(figsize=(8, 4))

axes = sns.lineplot(x=range(0, 4), y=bj_temps, 
                   marker="o", markersize=8, linewidth=2.5,
                   color='steelblue')

axes.set_xticks(range(0, 4))
axes.set_xticklabels(bj_hours)

axes.set_title('24小时温度变化趋势', fontsize=14, fontweight='bold', pad=20)
axes.set_xlabel('时间', fontsize=12)
axes.set_ylabel('温度 (°C)', fontsize=12)

for i, temp in enumerate(bj_temps):
    axes.annotate(f'{temp}°C', 
                 xy=(i, temp), 
                 xytext=(0, 10), 
                 textcoords='offset points',
                 ha='center', va='bottom',
                 fontsize=10, fontweight='bold',
                 bbox=dict(boxstyle='round,pad=0.3', 
                          facecolor='yellow', alpha=0.7))

axes.set_ylim(20, 31)

axes.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

legals = pd.read_csv("法人人数年度数据.csv", encoding='GB18030')
temp_legals = legals[1:11]

year_columns = [f"{x}年" for x in range(2009, 2018)]
new_legals = pd.melt(temp_legals, 
                     id_vars=['指标'], 
                     value_vars=year_columns,
                     var_name='年份', 
                     value_name='数量')

new_legals['数量'] = pd.to_numeric(new_legals['数量'], errors='coerce')

plt.figure(figsize=(20, 8))
sns.barplot(x="年份", y="数量", hue="指标", data=new_legals)

plt.title('法人人数年度变化趋势', fontsize=16, fontweight='bold')
plt.xlabel('年份', fontsize=12)
plt.ylabel('数量', fontsize=12)
plt.legend(ncol=4, loc='upper left')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
